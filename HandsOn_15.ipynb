Dijkstra's Algorithm 

```python
import heapq

def dijkstra(graph, source):
    dist = {v: float('inf') for v in graph}
    prev = {v: None for v in graph}
    dist[source] = 0
    pq = [(0, source)]
    while pq:
        d_u, u = heapq.heappop(pq)
        if d_u > dist[u]:
            continue
        for v, w in graph[u].items():
            alt = dist[u] + w
            if alt < dist[v]:
                dist[v] = alt
                prev[v] = u
                heapq.heappush(pq, (alt, v))
    return dist, prev

def reconstruct_path(prev, target):
    path = []
    u = target
    while u is not None:
        path.append(u)
        u = prev[u]
    return path[::-1]

if __name__ == "__main__":
    graph = {
        's': {'t': 10, 'y': 5},
        't': {'x': 1, 'y': 2},
        'x': {'z': 4},
        'y': {'t': 3, 'x': 9, 'z': 2},
        'z': {'s': 7, 'x': 6}
    }
    dist, prev = dijkstra(graph, 's')
    print("Vertex  Distance  Shortest Path")
    print("------  --------  -------------")
    for v in sorted(graph):
        path = reconstruct_path(prev, v)
        print(f"{v:>6}  {dist[v]:>8.0f}  {path}")
```

Bellman-Ford Algorithm


```python
def bellman_ford(graph, source):
    dist = {v: float('inf') for v in graph}
    prev = {v: None for v in graph}
    dist[source] = 0
    for _ in range(len(graph) - 1):
        for u in graph:
            if dist[u] != float('inf'):
                for v, w in graph[u].items():
                    alt = dist[u] + w
                    if alt < dist[v]:
                        dist[v] = alt
                        prev[v] = u
    for u in graph:
        if dist[u] != float('inf'):
            for v, w in graph[u].items():
                if dist[u] + w < dist[v]:
                    raise ValueError("Graph contains a negative-weight cycle")
    return dist, prev

def reconstruct_path(prev, target):
    path = []
    u = target
    while u is not None:
        path.append(u)
        u = prev[u]
    return path[::-1]

if __name__ == "__main__":
    graph_bf = {
        's': {'t': 6,  'y': 7},
        't': {'x': 5,  'z': -4, 'y': 8},
        'x': {'t': -2},
        'y': {'x': -3, 'z': 9},
        'z': {'s': 2,  'x': 7}
    }
    dist, prev = bellman_ford(graph_bf, 's')
    print("Vertex  Distance  Shortest Path")
    print("------  --------  -------------")
    for v in sorted(graph_bf):
        path = reconstruct_path(prev, v)
        print(f"{v:>6}  {dist[v]:>8.0f}  {path}")
```

Floyd-Warshall Algorithm 


```python
INF = float('inf')

def floyd_warshall(D):
    n = len(D)
    dist = [row.copy() for row in D]
    for k in range(n):
        for i in range(n):
            for j in range(n):
                if dist[i][k] + dist[k][j] < dist[i][j]:
                    dist[i][j] = dist[i][k] + dist[k][j]
    return dist

if __name__ == "__main__":
    D0 = [
        [0,   3,   INF, 7],
        [8,   0,   2,   INF],
        [5,   INF, 0,   1],
        [2,   INF, INF, 0]
    ]
    D = floyd_warshall(D0)
    for row in D:
        print(row)
```
